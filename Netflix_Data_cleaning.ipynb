```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("netflix_titles.csv")
```

```python
df
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
df.info()
```

```python
df["director"].fillna("unknown",inplace=True)
```

```python
df.isnull().sum()
```

```python
df["cast"].fillna("unknown",inplace=True)
df["country"].fillna("unknown",inplace=True)
df["rating"].fillna("unknown",inplace=True)
df["duration"].fillna("unknown",inplace=True)
df.isnull().sum()
```

```python
df['date_added'].mode()
```

```python
df.describe(include=object)
```

```python

df["date_added"].fillna("January 1, 2020",inplace=True)
df.isnull().sum()
```

```python
df.drop_duplicates(inplace=True)
df.shape
```

```python
df
```

```python
df["rating"].unique()
```

```python
df["rating"]=df["rating"].replace("74 min","R")
df["rating"]=df["rating"].replace("84 min","R")
df["rating"]=df["rating"].replace("66 min","R")
df
```
